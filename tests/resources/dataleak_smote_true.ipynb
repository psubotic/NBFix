```python
import numpy as np
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split

# random data:
X = np.genfromtxt("data.csv", delimiter=",", dtype=None)
y = np.random.choice(2, size=10)
```

```python
# resample the WHOLE dataset before splitting - a real leak: synthetic
# samples can be generated from neighbors that end up split across
# train and test
smote = SMOTE()
X_res, y_res = smote.fit_resample(X, y)
```

```python
# then split
X_train, X_test, y_train, y_test = train_test_split(X_res, y_res, test_size=0.25, random_state=2)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
a = lr.fit(X_train, y_train)
```

```python
y_pred = lr.predict(X_test)
```
