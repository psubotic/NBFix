```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

X = pd.read_excel("data.xlsx")
y = X
```

```python
# taint the WHOLE dataset before splitting
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.25, random_state=2)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
a = lr.fit(X_train, y_train)
```

```python
# .score() as the test-usage sink, instead of .predict()
acc = lr.score(X_test, y_test)
```
